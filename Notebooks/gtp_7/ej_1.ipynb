Implemente un **algoritmo de optimización por enjambre de partı́culas** y utilı́celo para encontrar el mı́nimo global de las funciones del Ejercicio 1 de la Guı́a de trabajos prácticos 6. 
Compare los resultados en relación a los obtenidos con algoritmos genéticos, en términos de las soluciones encontradas y la velocidad de convergencia.

### Función I.


```python
def funcion(x):
    return -x * np.sin(np.sqrt(np.abs(x)))

def func_error(x):
    return funcion(x)
```

```python
import numpy as np
import random

N = 100
D = 1
lim1 = -512
lim2 = 512

x = np.random.random((N,D))
x[:,0] = x[:,0] * (abs(lim1)+abs(lim2)) + lim1
v = np.random.random((N,D))

yk = x
yglob = x[1,:]

while func_error(yglob)>=-418:
    for k in range(N):
        if func_error(x[k,])<func_error(yk[k,]): 
            yk[k,] = x[k,]

        if func_error(yk[k,])<func_error(yglob): 
            yglob = yk[k,]

    for k in range(N):
        for i in range(x.shape[1]):
            v[k,i] += random.random()*(yk[k,i]-x[k,i]) + random.random()*(yglob[i]-x[k,i])
            
    x = np.clip(x + v,lim1,lim2) 

print(yglob)
        
```

### Función II.


```python
def funcion_b(X):
    x = X[0]
    y = X[1]
    return ((x**2+y**2)**0.25) * ((np.sin(50*((x**2+y**2)**0.1))**2) + 1)

def func_error_b(x):
    return funcion_b(x)
```

```python
import numpy as np
import random

N = 100
D = 2
lim1 = -100
lim2 = 100

x = np.random.random((N,D))
x[:,:] = x[:,:] * (abs(lim1)+abs(lim2)) + lim1
v = np.random.random((N,D))

yk = x
yglob = x[1,:]
it = 0
maxIter = 500

while func_error_b(yglob) > 0.3 and it < maxIter:
    for k in range(N):
        if func_error_b(x[k,])<func_error_b(yk[k,]): 
            yk[k,] = x[k,]
        if func_error_b(yk[k,])<func_error_b(yglob): 
            yglob = yk[k,]
    for k in range(N):
        for i in range(x.shape[1]):
            v[k,i] += random.random()*(yk[k,i]-x[k,i]) + random.random()*(yglob[i]-x[k,i])
    x = np.clip(x + v,lim1,lim2) 
    #print(f"{yglob}, f: {funcion(yglob)}")

    it += 1

print(f"Mejor par (x, y) encontrado: {yglob}, f(x,y) = {funcion_b(yglob):.4f}")
print(f"Epoca final: {it}\n")
        
```
