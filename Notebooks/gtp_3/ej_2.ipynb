## Ejercicio 2.
Mediante el esquema de cinco particiones generadas con KFold, compare el desempeño del perceptrón multicapa con los siguientes clasificadores:

- Naive Bayes
- Análisis Discriminante Lineal
- K vecinos más cercanos
- Árbol de decisión
- Máquina de Soporte Vectorial


```python
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score
from sklearn.neural_network import MLPClassifier
import numpy as np
from sklearn import datasets
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn import tree
from sklearn import svm

X, y = datasets.load_digits(return_X_y=True)

mlp = MLPClassifier(hidden_layer_sizes=(64,32), activation='logistic', learning_rate_init=0.01, max_iter=900, random_state=42)
gaussian_nb = GaussianNB()
kneighbors = KNeighborsClassifier(n_neighbors=3)
adl = LinearDiscriminantAnalysis()
arbol = tree.DecisionTreeClassifier()
vector_machine = svm.SVC()


kf = KFold(n_splits=5, shuffle=True, random_state=42)

acc_mlp = []
acc_gb = []
acc_knn = []
acc_adl = []
acc_tree = []
acc_svm = []
    
for i, (train_index, test_index) in enumerate(kf.split(X)):
    X_train_kf, X_test_kf = X[train_index], X[test_index]
    y_train_kf, y_test_kf = y[train_index], y[test_index]

    # MLP
    mlp.fit(X_train_kf, y_train_kf)
    y_pred_mlp = mlp.predict(X_test_kf)
    accuracy_kf_mlp = accuracy_score(y_test_kf, y_pred_mlp)

    # Naive Bayes
    y_pred_gb = gaussian_nb.fit(X_train_kf, y_train_kf).predict(X_test_kf)
    accuracy_kf_gb = accuracy_score(y_test_kf, y_pred_gb)

    # KNeighbors
    kneighbors.fit(X_train_kf, y_train_kf)
    y_pred_knn = kneighbors.predict(X_test_kf)
    accuracy_kf_knn = accuracy_score(y_test_kf, y_pred_knn)

    # ADL
    adl.fit(X_train_kf, y_train_kf)
    y_pred_adl = adl.predict(X_test_kf)
    accuracy_kf_adl = accuracy_score(y_test_kf, y_pred_adl)

    # Decision Tree
    arbol.fit(X_train_kf, y_train_kf)
    y_pred_tree = arbol.predict(X_test_kf)
    accuracy_kf_tree = accuracy_score(y_test_kf, y_pred_tree)

    # SVM
    vector_machine.fit(X_train_kf, y_train_kf)
    y_pred_svm = vector_machine.predict(X_test_kf)
    accuracy_kf_svm = accuracy_score(y_test_kf, y_pred_svm)

    print(f"\nFold {i+1} Accuracy (MLP): {accuracy_kf_mlp}")
    print(f"Fold {i+1} Accuracy (Gaussian NB): {accuracy_kf_gb}")
    print(f"Fold {i+1} Accuracy (KNeighbors): {accuracy_kf_knn}")
    print(f"Fold {i+1} Accuracy (ADL): {accuracy_kf_adl}")
    print(f"Fold {i+1} Accuracy (Decision Tree): {accuracy_kf_tree}")
    print(f"Fold {i+1} Accuracy (SVM): {accuracy_kf_svm}\n")

    acc_mlp.append(accuracy_kf_mlp)
    acc_gb.append(accuracy_kf_gb)
    acc_knn.append(accuracy_kf_knn)
    acc_adl.append(accuracy_kf_adl)
    acc_tree.append(accuracy_kf_tree)
    acc_svm.append(accuracy_kf_svm)

print("\n")
print("MLP Mean Accuracy:", np.mean(acc_mlp))
print("MLP Standard Deviation:", np.std(acc_mlp))
print("Gaussian NB Mean Accuracy:", np.mean(acc_gb))
print("Gaussian NB Standard Deviation:", np.std(acc_gb))
print("KNeighbors Mean Accuracy:", np.mean(acc_knn))
print("KNeighbors Standard Deviation:", np.std(acc_knn))
print("ADL Mean Accuracy:", np.mean(acc_adl))
print("ADL Standard Deviation:", np.std(acc_adl))
print("Decision Tree Mean Accuracy:", np.mean(acc_tree))
print("Decision Tree Standard Deviation:", np.std(acc_tree))
print("SVM Mean Accuracy:", np.mean(acc_svm))
print("SVM Standard Deviation:", np.std(acc_svm))
```
