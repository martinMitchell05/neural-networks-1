## Ejercicio 2.
Mediante el esquema de cinco particiones generadas con KFold, compare el desempeño del perceptrón multicapa con los siguientes clasificadores:

- Naive Bayes
- Análisis Discriminante Lineal
- K vecinos más cercanos
- Árbol de decisión
- Máquina de Soporte Vectorial


```python
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score
from sklearn.neural_network import MLPClassifier
import numpy as np
from sklearn import datasets
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, StandardScaler
from sklearn import tree
from sklearn import svm

X, y = datasets.load_digits(return_X_y=True)

mlp = MLPClassifier(hidden_layer_sizes=(64,32), activation='logistic', learning_rate_init=0.01, max_iter=900, random_state=42)
gaussian_nb = GaussianNB()
kneighbors = KNeighborsClassifier(n_neighbors=3, weights='distance', algorithm='auto', p=2, metric='minkowski')
adl = LinearDiscriminantAnalysis()
arbol = tree.DecisionTreeClassifier(criterion='gini', random_state=42)
vector_machine = svm.SVC(kernel='sigmoid', random_state=42)

X_scaler = StandardScaler()
    
kf = KFold(n_splits=5, shuffle=True, random_state=42)

acc_mlp = []
acc_gb = []
acc_knn = []
acc_adl = []
acc_tree = []
acc_svm = []
    
for i, (i_trn, i_tst) in enumerate(kf.split(X)):
    X_trn, X_tst = X[i_trn], X[i_tst]
    y_trn, y_tst = y[i_trn], y[i_tst]

    X_trn = X_scaler.fit_transform(X_trn)
    X_tst = X_scaler.transform(X_tst)

    # MLP
    mlp.fit(X_trn, y_trn)
    y_pred_mlp = mlp.predict(X_tst)
    accuracy_kf_mlp = accuracy_score(y_tst, y_pred_mlp)

    # Naive Bayes
    y_pred_gb = gaussian_nb.fit(X_trn, y_trn).predict(X_tst)
    accuracy_kf_gb = accuracy_score(y_tst, y_pred_gb)

    # KNeighbors
    kneighbors.fit(X_trn, y_trn)
    y_pred_knn = kneighbors.predict(X_tst)
    accuracy_kf_knn = accuracy_score(y_tst, y_pred_knn)

    # ADL
    adl.fit(X_trn, y_trn)
    y_pred_adl = adl.predict(X_tst)
    accuracy_kf_adl = accuracy_score(y_tst, y_pred_adl)

    # Decision Tree
    arbol.fit(X_trn, y_trn)
    y_pred_tree = arbol.predict(X_tst)
    accuracy_kf_tree = accuracy_score(y_tst, y_pred_tree)

    # SVM
    vector_machine.fit(X_trn, y_trn)
    y_pred_svm = vector_machine.predict(X_tst)
    accuracy_kf_svm = accuracy_score(y_tst, y_pred_svm)

    print(f"\nFold {i+1} Accuracy (MLP): {accuracy_kf_mlp}")
    print(f"Fold {i+1} Accuracy (Gaussian NB): {accuracy_kf_gb}")
    print(f"Fold {i+1} Accuracy (KNeighbors): {accuracy_kf_knn}")
    print(f"Fold {i+1} Accuracy (ADL): {accuracy_kf_adl}")
    print(f"Fold {i+1} Accuracy (Decision Tree): {accuracy_kf_tree}")
    print(f"Fold {i+1} Accuracy (SVM): {accuracy_kf_svm}\n")

    acc_mlp.append(accuracy_kf_mlp)
    acc_gb.append(accuracy_kf_gb)
    acc_knn.append(accuracy_kf_knn)
    acc_adl.append(accuracy_kf_adl)
    acc_tree.append(accuracy_kf_tree)
    acc_svm.append(accuracy_kf_svm)

print("\n")
print(f"MLP Media de Acierto: {np.mean(acc_mlp)}")
print(f"MLP Desviación Estándar: {np.std(acc_mlp)}\n")
print(f"Gaussian NB Media de Acierto: {np.mean(acc_gb)}")
print(f"Gaussian NB Desviación Estándar: {np.std(acc_gb)}\n")
print(f"KNeighbors Media de Acierto: {np.mean(acc_knn)}")
print(f"KNeighbors Desviación Estándar: {np.std(acc_knn)}\n")
print(f"ADL Media de Acierto: {np.mean(acc_adl)}")
print(f"ADL Desviación Estándar: {np.std(acc_adl)}\n")
print(f"Decision Tree Media de Acierto: {np.mean(acc_tree)}")
print(f"Decision Tree Desviación Estándar: {np.std(acc_tree)}\n")
print(f"SVM Media de Acierto: {np.mean(acc_svm)}")
print(f"SVM Desviación Estándar: {np.std(acc_svm)}")
```
