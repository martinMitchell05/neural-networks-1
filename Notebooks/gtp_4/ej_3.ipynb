```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans

tabla = pd.read_csv('../../Data/gtp_2/iris81_trn.csv', header=None).to_numpy()
X = tabla[:,0:4]

valores_k = range(2, 11)
compactitud = []

for k in valores_k:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    
    compactitud.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(valores_k, compactitud, marker='o', linestyle='-', color='b')
plt.title('Metodo del Codo')
plt.xlabel('k (numero de clusters)')
plt.ylabel('Compactitud')
plt.xticks(valores_k)
plt.grid(True)
plt.show()
```
