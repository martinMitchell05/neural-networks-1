## Ejercicio 1.

Implemente el algoritmo de entrenamiento de un SOM bidimensional de forma que se pueda ver gráficamente el mapa topológico durante todo el proceso.
Para poder observar el ordenamiento topológico, en el mapa grafique lı́neas de unión entre pares de neuronas vecinas.

Realice las pruebas de ordenamiento topológico con los patrones provistos en los archivos circulo.csv y te.csv, usando un máximo de 1000 épocas de entrenamiento. Dichos datos fueron generados aleatoriamente con distribución uniforme dentro de un cı́rculo de radio 1 centrado en el origen y una T como muestra la figura.

#### Parte 2.
Para cada distribución grafique los datos de entrenamiento coloreando cada punto según la neurona ganadora correspondiente, y agregue el centroide de cada neurona con otro marcador.

#### Parte 3.

### Inicialización y entrenamiento del SOM

**Entrenamiento para "circulo"**

```python
%matplotlib qt
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

X = pd.read_csv('../../Data/gtp_4/circulo.csv', header=None).to_numpy()
W = np.random.random((10, 10, X.shape[1])) - 0.5 

epoca = 1
epoca_max = 600
Areas = [5, 4, 3, 2, 1]
Betas = [0.8, 0.6, 0.5, 0.3, 0.0]

plt.ion()
fig, ax = plt.subplots(figsize=(8,8))

while epoca < epoca_max:

    if epoca < 100:
        A_n = Areas[0]
        beta_n = Betas[0]
        tasa_actual = 0.8
    elif 100 <= epoca < 300:
        A_n = Areas[1]
        beta_n = Betas[1]
        tasa_actual = 0.6
    elif 300 <= epoca < 450:
        A_n = Areas[2]
        beta_n = Betas[2]
        tasa_actual = 0.4
    elif 450 <= epoca < 550:
        A_n = Areas[3]
        beta_n = Betas[3]
        tasa_actual = 0.2
    elif 550 <= epoca < epoca_max:
        A_n = Areas[4] 
        beta_n = Betas[4]
        tasa_actual = 0.1

    color_entr_por_neur = np.zeros(X.shape[0])

    for n in range(X.shape[0]):
        # buscar neurona ganadora 
        distancias = np.linalg.norm(X[n] - W, axis=2) 
        j_ganador = np.unravel_index(np.argmin(distancias), distancias.shape) 

        color_entr_por_neur[n] = np.argmin(distancias)

        for i in range(W.shape[0]):
            for z in range(W.shape[1]):
                
                # ver la distancia de las otras neuronas a la ganadora
                neurona_i = np.array([i, z])
                neurona_ganadora = np.array(j_ganador)
                d = np.linalg.norm(neurona_i - neurona_ganadora)

                # actualizar los pesos
                if d == 0: 
                    W[i, z] += tasa_actual * (X[n] - W[i, z])
                
                elif d <= A_n: 
                    W[i, z] += tasa_actual * beta_n * (X[n] - W[i, z])


    if epoca % 10 == 0 or epoca == 1:
        ax.clear()
        ax.grid(True, linestyle=':', alpha=0.5)
        ax.set_xlabel("x1")
        ax.set_ylabel("x2")
        ax.set_title(f"Época: {epoca} | Área: {A_n} | Tasa: {tasa_actual:.1f}", loc='center')

        # Se usa 'cmap' para asignar un color distinto a cada número de neurona, c='' asigna los colores en función de esos valores
        ax.scatter(X[:,0], X[:,1], c=color_entr_por_neur, cmap='hsv', marker='o', alpha=0.6, s=25)

        # Dibujar el mapa (uniones entre neuronas)
        for i in range(W.shape[0]):
            ax.plot(W[i, :, 0], W[i, :, 1], color='gray', linewidth=1.0, alpha=0.4)
        for z in range(W.shape[1]):
            ax.plot(W[:, z, 0], W[:, z, 1], color='gray', linewidth=1.0, alpha=0.4)


        # Sacar una dimension a W temporalmente para que scatter lo lea de una sola vez
        W_plano = W.reshape(-1, X.shape[1])
        ax.scatter(W_plano[:,0], W_plano[:,1], facecolors='blue', edgecolors='black', marker='X', s=30, zorder=3)

        plt.pause(0.1)

    epoca += 1

plt.ioff()
plt.show()
```

```python
%matplotlib qt
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

X = pd.read_csv('../../Data/gtp_4/te.csv', header=None).to_numpy()
W = np.random.random((100, X.shape[1])) - 0.5 

epoca = 1
epoca_max = 600

Areas = [5, 4, 3, 2, 1]
Betas = [0.8, 0.6, 0.5, 0.3, 0.0]

plt.ion()
fig, ax = plt.subplots(figsize=(8,8))

while epoca < epoca_max:

    if epoca < 100:
        A_n = Areas[0]
        beta_n = Betas[0]
        tasa_actual = 0.8
    elif 100 <= epoca < 300:
        A_n = Areas[1]
        beta_n = Betas[1]
        tasa_actual = 0.6
    elif 300 <= epoca < 450:
        A_n = Areas[2]
        beta_n = Betas[2]
        tasa_actual = 0.4
    elif 450 <= epoca < 550:
        A_n = Areas[3]
        beta_n = Betas[3]
        tasa_actual = 0.2
    elif 550 <= epoca < epoca_max:
        A_n = Areas[4] 
        beta_n = Betas[4]
        tasa_actual = 0.1

    color_entr_por_neur = np.zeros(X.shape[0])

    for n in range(X.shape[0]):
        # buscar neurona ganadora 
        distancias = np.linalg.norm(X[n] - W, axis=1) 
        j_ganador = np.argmin(distancias) 

        color_entr_por_neur[n] = np.argmin(distancias)

        for i in range(W.shape[0]):
            # ver la distancia de las otras neuronas a la ganadora
            d = np.abs(i - j_ganador)

            # actualizar los pesos
            if d == 0: 
                W[i] += tasa_actual * (X[n] - W[i])
            elif d <= A_n: 
                W[i] += tasa_actual * beta_n * (X[n] - W[i])


    if epoca % 1 == 0 or epoca == 1:
        ax.clear()
        ax.grid(True, linestyle=':', alpha=0.5)
        ax.set_xlabel("x1")
        ax.set_ylabel("x2")
        ax.set_title(f"Época: {epoca} | Área: {A_n} | Tasa: {tasa_actual:.1f}", loc='center')

        # Se usa 'cmap' para asignar un color distinto a cada número de neurona, c='' asigna los colores en función de esos valores
        ax.scatter(X[:,0], X[:,1], c=color_entr_por_neur, cmap='hsv', marker='o', alpha=0.6, s=25)

        # Dibujar el mapa 1D
        ax.plot(W[:, 0], W[:, 1], color='gray', linewidth=1.0, alpha=0.4)
        ax.scatter(W[:,0], W[:,1], facecolors='blue', edgecolors='black', marker='X', s=30, zorder=3)

        plt.pause(0.1)

    epoca += 1

plt.ioff()
plt.show()
```
