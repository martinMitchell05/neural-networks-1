### Algoritmo de entrenamiento: Perceptrón simple

```
1. Iniciar pesos

Mientras (epoca < epoca_maxima):

    # Entrenar ...
    Por cada par entradas/salida_esperada:
        2. Calcular salida 'y' (Producto interno)
        3. Ajustar pesos (en este caso no se verifica si hubo error o no)
        4. Repetir desde 2

    # Comprobar entrenamiento ...
    Por cada par entradas/salida_esperada:
        1. Calcular salida 'y' (Producto interno)
        2. Comparar con 'salida_esperada'
        3. Sumar los aciertos
        4. Repetir
        
    Calcular una 'tasa de aciertos' como método de corte
    Si la tasa es mayor a (ej. 90%): Salir; Sino: aumentar epoca y repetir
    
```

````python
import numpy as np
import pandas as pd
import random

cabeceras = ['x1', 'x2', 'd']

df_trn = pd.read_csv('../../Data/gtp-1/OR_trn.csv', header=None, names=cabeceras)
df_tst = pd.read_csv('../../Data/gtp-1/OR_tst.csv', header=None, names=cabeceras)

# print(df_trn.describe())

print("\n")

x1_trn = df_trn['x1'].to_numpy()
x2_trn = df_trn['x2'].to_numpy()
d_trn = df_trn['d'].to_numpy()

epoc_max = 500
epoc_actual = 1
mu = 0.01

x = np.array([x1_trn, x2_trn, -np.ones(d_trn.size)])
print(x.shape)

# w_aux = random.random() - 0.5
# w = np.array([w_aux, w_aux, w_aux])
w = np.array([random.random()-0.5, random.random()-0.5, random.random()-0.5])
print(w.shape)

while epoc_actual < epoc_max:

    acierto = 0

    # Entrenar:
    for i in range(0, x1_trn.size):
        y = np.sign(np.dot(x[:, i], w[:]))

        error = mu/2 * (d_trn[i] - y)

        w = w + error * x[:, i]

    # Verificar:
    for i in range(0, d_trn.size):

        y_actualizada = np.sign(np.dot(x[:,i], w[:]))
        if (y_actualizada == d_trn[i]):
            acierto += 1


    tasa_aciertos = acierto/d_trn.size
    print(f"Tasa aciertos = {tasa_aciertos}, Errores = {d_trn.size-acierto}")

    if (tasa_aciertos > 0.99):
        print(f"\nTasa de acierto = {tasa_aciertos}, Epoca: {epoc_actual}")
        print(f"Pesos finales = {w}")
        break
    else:
        epoc_actual += 1


print(f"Epoca final: {epoc_actual}")    

````

#### Parte 2: Verificar con nuevos datos
